**Data Preprocessing**


Feature scaling: we will bring all the numbers to a standard scale by dividing them with 255.

```python
xtrain_scl=xtrain.astype("float32")/255
xtest_scl=xtest.astype("float32")/255
```

```python
from keras.utils import to_categorical
ytrain_scl=to_categorical(ytrain)
ytest_scl=to_categorical(ytest)
```

```python
ytrain[0]
```

```python
ytrain_scl[0]
```

**expand the dimensions to tell keras about the color of the image**


1.RGB - we dont expand dimensions for RGB as it is mentioned in the shape of colored images.



2.Grayscale - needs to be done for grayscale as it is NOT mentioned in the shape of grayscale images

```python
xtrain_scl.shape
```

```python
import numpy as np
xtrain_scl = np.expand_dims(xtrain_scl,axis=-1)
xtest_scl = np.expand_dims(xtest_scl,axis=-1)
```

```python
xtrain_scl.shape
```

### **Model Building**

```python
ytrain_scl.shape
```

```python
ytrain_scl.shape[1]
```

```python
from keras.models import Sequential
from keras.layers import Dense,Input,Dropout,Flatten,Conv2D,MaxPool2D

##Initialize the model
model=Sequential()

## Add the input layer
model.add(Input(shape=xtrain_scl.shape[1:],))

##Add the convolution layer
model.add(Conv2D(32,kernel_size=(3,3),activation='relu'))

##Add another pooling layer
model.add(MaxPool2D(pool_size=(2,2)))

##Add another convolution layer
model.add(Conv2D(64,kernel_size=(3,3),activation='relu'))
# Add another pooling layer
model.add(MaxPool2D(pool_size=(2,2)))

##Add flatten layer
model.add(Flatten())

##Add Dropout Layer
model.add(Dropout(0.3))

#Add a Hidden Layer
model.add(Dense(128,activation='relu'))

#Add the output layer
model.add(Dense(units=ytrain_scl.shape[1],activation='softmax'))

#summary
model.summary()
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
from keras.callbacks import EarlyStopping
es=EarlyStopping(monitor='val_loss',patience=5,verbose=1)
```

```python
hist=model.fit(xtrain_scl,ytrain_scl,epochs=50,validation_split=0.2,callbacks=[es])
```

### **Learning Curve**

```python
import matplotlib.pyplot as plt
plt.plot(hist.history['loss'],label="Traning loss")
plt.plot(hist.history['val_loss'],label="Validation Loss")
plt.xlabel("Number of Epochs")
plt.ylabel("Categorical Loss")
plt.title("Learning curve")
plt.legend()
plt.show()
```

### **Model Evaluation**

```python
model.evaluate(xtrain_scl,ytrain_scl)
```

```python
model.evaluate(xtest_scl,ytest_scl)
```

## **Define a method that takes image paths as inputs, and does the necessary preprocessing**

```python
#0-255
# model : [0,15,255]
# image : [255,240,0]

# subtract your content with 255 : [0,15,255]
def preprocess_image(img_path):
  img = cv2.imread(img_path,cv2.IMREAD_GRAYSCALE)
  # convert background and digits to black and white colors as the model has learnt
  img = 255-img
  # scaling of image data to bring all the pixel amount to a standard scale
  img_scl = img.astype('float32')/255
  # Reshaping of data
  img_scl = img_scl.reshape(1,28,28,1)
  return img_scl
```

```python
path1 = r"/content/Example_3.jpg"
path2 = r"/content/Example_7.jpg"
path3 = r"/content/Example_9.jpg"
```

```python
from keras.models import load_model
model_final = load_model(r"/content/handwrittendigitsmodel.keras")
```

```python
def predict(path):
  img_scl = preprocess_image(path)
  probs = model_final(img_scl)
  pred = np.argmax(probs)
  return pred
```
